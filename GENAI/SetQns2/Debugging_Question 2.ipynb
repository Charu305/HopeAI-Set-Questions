```python
import faiss
import numpy as np
import google.generativeai as genai

genai.configure(api_key="AIzaSyC2on_wIP287EV7-9LHHZISVMdAlGGVw3s")

def get_embedding(text):
    response = genai.embed_content(
        model="models/gemini-embedding-001",
        content=text,
        task_type="retrieval_document"
    )
    return np.array(response['embedding'], dtype='float32')

# Create FAISS index dynamically
sample = get_embedding("hello")
d = len(sample)
index = faiss.IndexFlatL2(d)

documents = [
    "AI is transforming industries",
    "FAISS is great for vector search",
    "Generative models are powerful"
]

doc_embeddings = np.vstack([get_embedding(doc) for doc in documents]).astype("float32")
index.add(doc_embeddings)

query = "How is AI transforming industries?"
query_vector = np.array([get_embedding(query)], dtype="float32")

_, indices = index.search(query_vector, k=1)
retrieved_doc = documents[indices[0][0]]

print("Retrieved:", retrieved_doc)
```
